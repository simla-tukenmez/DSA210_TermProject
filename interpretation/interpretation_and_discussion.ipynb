## The Machine Learning Reality Check

Beyond hypothesis testing, we attempted to predict daily volatility using Machine Learning models (Random Forest, Ridge, Lasso).

### The "Negative R²" Finding
Contrary to expectations, complex ML models failed to outperform simple statistical baselines, yielding negative $R^2$ scores in the test set. This is a crucial finding regarding market efficiency in crisis periods.

**Why did ML fail? (The Diagnosis)**

1.  **Low Signal-to-Noise Ratio:**
    * Financial markets during crisis (earthquakes, elections) are driven by exogenous shocks rather than learnable historical patterns.
2.  **Fat Tails (Black Swans):**
    * Our EDA revealed a kurtosis of **3.66**. Extreme events happen far more frequently than Gaussian models expect, causing MSE-based loss functions to fail.
3.  **Structural Breaks (Concept Drift):**
    * The rapid shift from a low-interest regime (8.5%) to a high-interest regime (50%) meant that patterns learned in 2023 were invalid for 2024.

**Key Takeaway:** In high-volatility emerging markets, parsimonious statistical models (OLS) often outperform complex algorithms due to lower variance and better interpretability.

## 2. Deep Dive: Why Traditional Indicators Failed

### The Surprising Weakness of Traditional Models

One of the most striking findings is that **traditional economic indicators performed very poorly**:

```
Traditional Model R²: 0.26%
Digital Model R²:     1.87%
Ratio:                7.2X difference!
```

### Why Did This Happen?

#### **Reason 1: Crisis Context**
During extreme crises, **fundamental relationships break down**:
- Normal periods: Interest rates ↑ → BIST100 ↓ (predictable)
- Crisis periods: Panic dominates fundamentals
- Investors focus on survival, not valuation

#### **Reason 2: Policy Lag**
Traditional indicators (interest rates, inflation) are:
- Released monthly (not daily)
- Lagged (CPI reflects past inflation)
- Slow to adjust

Digital indicators (Google searches) are:
- Real-time
- Forward-looking (people search when worried about future)
- High-frequency

#### **Reason 3: Variable Selection**
We tested:
- USD/TRY **changes** (daily pct_change)
- Funding Cost **changes** (daily difference)

But perhaps **levels** or **volatility** matter more in crisis:
- USD/TRY level: 18 → 34 TRY (+89% depreciation!)
- This sustained pressure ≠ daily changes

### Academic Significance

This finding challenges conventional wisdom:
- Literature: Fundamentals drive markets
- Our finding: In crises, **psychology** > fundamentals

**Implication:** Portfolio managers should monitor Google Trends during crisis periods!

## 4. Deep Dive: Why Sentiment Failed (H5)

### The Disappointing Result

```
H5: Sentiment Prediction Test
Same-day correlation:  r = 0.171 (moderate)
Leading correlation:   r = 0.003 (zero!)
Regression p-value:    0.9318 (not significant)
```

### Why Doesn't Sentiment Predict Returns?

#### **Theory vs. Reality**

**Theory said:**
- Negative sentiment → People expect market drop → Sell pressure → Returns ↓

**Reality showed:**
- Market drops → People complain on Ekşi Sözlük → Negative sentiment

#### **Explanation: Retail vs. Institutional**

Ekşi Sözlük users are **retail investors**:
- React to market moves (don't drive them)
- Complain after losses (catharsis)
- Limited trading capital

**Institutional investors** drive actual returns:
- They don't post on Ekşi Sözlük
- They trade on fundamentals + technical signals
- Sentiment = noise for them

#### **Data Lag Issue**

```
Day 1 (Feb 7):  BIST100 crashes -9.01%
Day 1 evening:  People write negative entries on Ekşi
Day 2 (Feb 8):  Market closed (no return to predict)
```

→ Sentiment lag creates measurement issues

### Academic Insight

**Literature context:**
- Twitter sentiment predicts US stocks (Bollen et al., 2011)
- Reddit WallStreetBets drives meme stocks

**Our finding:**
- Turkish social media (Ekşi Sözlük) does NOT predict
- Platform difference? (Forum vs. Twitter)
- Market structure difference? (Retail power in Turkey)

### Future Work

Test with:

- Intraday data (hourly sentiment → hourly returns)

### Final Thought

While our findings are robust, the following limitations should be noted:

* **Sentiment Lag:** We found social media sentiment to be *reactive* rather than *predictive*. Investors tend to complain on Ekşi Sözlük *after* the market drops, limiting its use as an early warning signal.
* **Google Trends Resolution:** Data is normalized on a 0-100 scale, which masks the absolute volume of searches.
* **Sample Size:** The 2.5-year window (N=701) is relatively short for training deep learning models (LSTMs), which contributed to the underperformance of complex models.

In an age of information abundance, **what people search reveals their fears**. And in financial markets, **fear drives volatility** more than any economic fundamental.
